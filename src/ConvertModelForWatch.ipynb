```python
EXERCISE_CLASSIFIER = 'Exercise'
```

```python
from utils import load_classifier

exercise_classifier = load_classifier(EXERCISE_CLASSIFIER)
print(type(exercise_classifier)) 
```

```python
import onnxmltools
import numpy as np
from skl2onnx.common.data_types import FloatTensorType

# Define the input shape (number of features)
num_features = exercise_classifier.n_features_in_  # Get the number of input features

# Define the input type
initial_type = [("input", FloatTensorType([None, num_features]))]  # Batch size is "None"

# Convert XGBoost model to ONNX
onnx_model = onnxmltools.convert_xgboost(exercise_classifier, initial_types=initial_type, name="XGBoostClassifier")

# Save the ONNX model
onnxmltools.utils.save_model(onnx_model, "classifier.onnx")
print("Model converted to ONNX successfully!")
```
